```python
import torch

path = "mistral_two_senators_activations.pt"
data = torch.load(path)

print("num records:", len(data))
print("type:", type(data))
print("keys:", data[0].keys())
```

```python
for i, rec in enumerate(data):
    print(f"\nRecord {i}: {rec['senator']}")
    print("input_ids shape:", rec["input_ids"].shape)              # usually [1, T]
    print("attention_mask shape:", rec["attention_mask"].shape)    # usually [1, T]
    print("layer_hidden_avg shape:", rec["layer_hidden_avg"].shape) # should be [32, 4096]
    print("head_avg shape:", rec["head_avg"].shape)                 # should be [32, 32, 128]
```

```python
for i, rec in enumerate(data):
    lh = rec["layer_hidden_avg"]
    hh = rec["head_avg"]

    print(f"\nRecord {i}: {rec['senator']}")
    print("layer_hidden_avg has nan:", torch.isnan(lh).any().item())
    print("layer_hidden_avg has inf:", torch.isinf(lh).any().item())
    print("head_avg has nan:", torch.isnan(hh).any().item())
    print("head_avg has inf:", torch.isinf(hh).any().item())
```

```python
if len(data) >= 2:
    a = data[0]["head_avg"].flatten()
    b = data[1]["head_avg"].flatten()

    cos = torch.nn.functional.cosine_similarity(a, b, dim=0)
    print("cosine similarity between first two head_avg:", cos.item())

    diff = (a - b).abs().mean()
    print("mean abs diff:", diff.item())
```

```python
for rec in data:
    print(rec["senator"], "->", rec["prompt"])
```

```python
import torch
import matplotlib.pyplot as plt

# load your saved pt file
data = torch.load("mistral_two_senators_activations.pt")

# pick one record, e.g. Bernie Sanders
record = data[0]

# shape: [32, 32, 128]
head_avg = record["head_avg"]

# reduce 128-d vector -> 1 scalar per (layer, head)
# option 1: L2 norm
heatmap = torch.norm(head_avg, dim=-1)   # shape: [32, 32]

# plot
plt.figure(figsize=(10, 8))
plt.imshow(heatmap.cpu().numpy(), aspect="auto", origin="lower")
plt.colorbar(label="L2 norm of head_avg")
plt.xlabel("Head")
plt.ylabel("Layer")
plt.xticks(range(32))
plt.yticks(range(32))
plt.show()
```
